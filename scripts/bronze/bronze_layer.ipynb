### Ingest Files to Bronze Layer

```python
for item in INGESTION_CONFIG:
    print(f'Ingesting Data from {item['source']} -> to workspace.bronze.{item['table']}')

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(item["path"])
    )

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(f"coptic_analytics.bronze.{item['table']}")
    )
```
