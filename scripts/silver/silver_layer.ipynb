# Building Silver Layer and Applying Transformation:
- Standard column names
- Trimming white spaces
- Convert datatypes 

### Transforming **`Activity`** table

```python
%sql

CREATE OR REPLACE TABLE coptic_analytics.silver.activity
USING DELTA
AS

select 
    activity_id,
    church_id,
    trim(activity_name) as activity_name,
    is_mandatory,
    cast(weight as decimal(5, 2)) as weight
from coptic_analytics.bronze.activity
where deleted_at is null
```

### Transforming **`app_role`** table

```python
%sql

CREATE OR REPLACE TABLE coptic_analytics.silver.app_role
USING DELTA
AS

select 
    trim(role_name) as role_name, 
    trim(scope_level) as scope_level 
from coptic_analytics.bronze.app_role
```

### Transforming **`app_user`** table

```python
%sql

CREATE OR REPLACE TABLE coptic_analytics.silver.app_user
USING DELTA
AS
select
    user_id, 
    trim(role_name) as role_name, 
    diocese_id, 
    church_id, 
    department_id, 
    class_id, 
    member_id, 
    trim(email) as email
from coptic_analytics.bronze.app_user
where deleted_at is null
```
